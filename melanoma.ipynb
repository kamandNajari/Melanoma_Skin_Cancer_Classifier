```python
import cv2 
import tkinter as tk 
from tkinter import filedialog 
from ultralytics import YOLO

model = YOLO("best.pt")

root = tk.Tk() 
root.withdraw() 
image_path = filedialog.askopenfilename( 
    title="Select a skin lesion image", 
    filetypes=[("Image files", "*.jpg *.jpeg *.png *.bmp")] 
) 
root.destroy()

if not image_path:
    print("No image selected.") 
else:
    results = model(image_path)
    probs = results[0].probs
    class_names = results[0].names
    predicted_class = class_names[probs.top1]
    confidence = probs.top1conf.item() * 100

    image = cv2.imread(image_path)
    image = cv2.resize(image, (500, 500))
    label = f"{predicted_class.upper()} ({confidence:.1f}%)"
    color = (0, 0, 255) if predicted_class.lower() == "malignant" else (0, 200, 0)
    cv2.putText(image, label, (20, 40), cv2.FONT_HERSHEY_SIMPLEX, 1, color, 2)
    cv2.putText(image, "Educational use only - not a medical diagnosis", (20, 470),
               cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 1)
    cv2.imshow("Melanoma Classifier Result", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
